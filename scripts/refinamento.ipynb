```python
import duckdb
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
df = con.execute("SELECT * FROM produtos").df()
df.head(10)
```

```python
df2 = df.drop(columns=['id_categoria', 'id_fornecedor'])
df2.head(10)
```

```python
con.execute('''
CREATE TABLE IF NOT EXISTS dim_produtos(
    id_produto      BIGINT  ,
    nm_produto      TEXT    ,
    vl_produto        FLOAT
)''')
```

```python
con.execute('''insert into dim_produtos select * from df2''')
```

```python
df_dim = con.execute("SELECT * FROM dim_produtos").df()
df_dim.head(10)
```
